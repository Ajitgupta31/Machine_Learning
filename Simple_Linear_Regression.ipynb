```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.metrics import accuracy_score,r2_score,mean_squared_error
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

print ("--Simple Linear Regression--")

# Generate sample data
np.random.seed(0)
x_linear = 2 * np.random.rand(100,1)
y_linear = 4 + 3 * x_linear + np.random.randn(100,1)

# Split data into training and testing sets
x_train_linear, x_test_linear, y_train_linear, y_test_linear = train_test_split(x_linear, y_linear, test_size=0.2, random_state=42)

# Create a linear Regression model
linear_reg_model = LinearRegression()

# Train model
linear_reg_model.fit(x_train_linear, y_train_linear)

# Make prediction
y_pred_linear = linear_reg_model.predict(x_test_linear)

# Evaluate the model
mse = mean_squared_error(y_test_linear, y_pred_linear)
r2 = r2_score(y_test_linear, y_pred_linear)

print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"R-squared (R2) Score: {r2:.2f}")

# Plot the results
plt.figure(figsize=(8, 6))
plt.scatter(x_test_linear, y_test_linear, color='blue', label='Actual Data')
plt.plot(x_test_linear, y_pred_linear, color='red', linewidth=2, label='Linear Regression Fit')
plt.title('Simple Linear Regression')
plt.xlabel('X (Feature)')
plt.ylabel('Y (Target)')
plt.legend()
plt.grid(True)
plt.show()
```
