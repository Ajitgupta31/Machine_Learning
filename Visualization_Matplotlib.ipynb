### Definitions of Matplotlib Graph Types:

*   **Linear Plot (Line Plot)**: Displays data points connected by straight line segments. Used to show trends over time or ordered categories.
*   **Scatter Plot**: Displays individual data points, typically between two numerical variables. Used to observe relationships or correlations between variables.
*   **Bar Plot**: Presents categorical data with rectangular bars, where the length of each bar is proportional to the value it represents. Used for comparing different categories.
*   **Stem Plot**: Similar to a line plot but shows data points as markers at the end of vertical lines (stems) extending from a baseline. Often used for discrete signals or data.
*   **Step Plot**: Creates a series of horizontal and vertical line segments to represent data. The plot changes its value abruptly at certain points, forming a staircase-like pattern. Useful for visualizing functions that are constant over intervals.
*   **Hist Plot (Histogram)**: Represents the distribution of numerical data. It groups data into bins and shows the frequency of data points within each bin as bars. Used to understand the underlying distribution of a dataset.
*   **Box Plot (Box-and-Whisker Plot)**: Displays the distribution of a dataset through quartiles. It shows the median, quartiles, and potential outliers. Useful for comparing distributions between groups and identifying skewness.
*   **Pie Plot (Pie Chart)**: A circular statistical graphic, divided into slices to illustrate numerical proportion. In a pie chart, the arc length of each slice is proportional to the quantity it represents. Used for showing parts of a whole.
*   **Fill_Between Plot**: Fills the area between two horizontal curves. Commonly used to highlight a specific range or show uncertainty/confidence intervals.
