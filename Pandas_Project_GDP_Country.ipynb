# APPLY THE LEARNING AND UNDERSTAND THE REAL

**uNIVARIATE ANALYSIS** : Plot the graph using 1 variable is called Univariate analysis.

How to detect Univariate base on graph -- distplot or displot

**BIVARIATE ANALYSIS** - Plot the graph using 2 variable is called Bivariate Analysis

**Multivariate Analysis** - Plot the graph using many variable is called multivariable
